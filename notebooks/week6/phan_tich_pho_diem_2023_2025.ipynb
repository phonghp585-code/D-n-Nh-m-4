```python
from pathlib import Path
import pandas as pd
from src.week6_score_distribution import run

file_2023 = Path('data/processed/week3/exam_2023.parquet')
file_2025 = Path('data/dataset_diem_thi_thpt_2025.csv')
output_dir = Path('outputs/score_distribution')

run(file_2023, file_2025, output_dir)
```

## 1. Bảng thống kê mô tả

Đọc mean cùng median để hiểu vị trí phân phối; đọc standard deviation cùng Q1–Q3 để hiểu độ phân tán. Skewness và excess kurtosis mô tả hình dạng, không thay thế việc xem biểu đồ.

```python
stats = pd.read_csv(output_dir / 'thong_ke_mo_ta_2023_2025.csv')
display(stats.round(3))
```

## 2. So sánh tỷ lệ ngưỡng điểm

Các tỷ lệ được tính trên điểm hợp lệ của từng môn-năm: điểm liệt (≤1), dưới trung bình (<5), từ 8 trở lên (≥8), từ 9 trở lên (≥9).

```python
threshold_cols = ['year', 'subject', 'n', 'pct_le_1', 'pct_lt_5', 'pct_ge_8', 'pct_ge_9']
display(stats[threshold_cols].round(2))
```

## 3. Xem biểu đồ đã tạo

Histogram so sánh tần suất theo khoảng điểm; KDE làm mượt mật độ và có thể che các đỉnh nhỏ; boxplot thể hiện median/IQR; violin plot cho biết thêm hình dạng mật độ. Luôn dùng cùng thang điểm 0–10.

```python
figures = sorted((output_dir / 'figures').glob('*.png'))
print(f'{len(figures)} biểu đồ được tạo')
for path in figures:
    print(path)
```
