## Ghi chú diễn giải
- Mean/median dịch chuyển xuống: phân phối điểm có xu hướng thấp hơn, xét theo môn.
- SD và IQR phản ánh độ phân tán; skewness phản ánh độ lệch; kurtosis ở đây là excess kurtosis.
- So sánh tỷ lệ ≤1, <5, ≥8, ≥9 trên mẫu điểm hợp lệ. Cần xem xét khác biệt đề thi và cấu trúc thí sinh giữa hai năm.
